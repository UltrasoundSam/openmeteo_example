This notebook gives some demonstration on how to use the MeteoForecast class in order to understand how to obtain weather prediction data from the Open-Meteo API. The API allows you to obtain a range of different weather variables, including temperature, wind speed, amount of cloud cover and precipitation, all of which can be accessed using this Python module. Other information - which totals an impressive 90 TB of data available from OpenMeteo - such as soil moisture, solar irradiance, dew points, etc, can be added in future if required. 

This rich source of data is perfect for training and refining your meteorological data science projects - so let's get started! The main class for accessing forecast data is the ```MeteoForecast``` class, so let's import it:

````python
from openmeteo_example import MeteoForecast
````

Now we have imported it, let's use it! In order to create the object, we need to provide it with the latitude and longitude of a particular location. According to the [API documentation](https://open-meteo.com/en/docs#hourly_parameter_definition), we should expect a spatial resolution of better than 10 km, which should be more than good enough for most purposes. 
Let's choose a location in the UK, and create the object: 

````python
# Define location
LOCATION = (53.22141, -4.20329)

# Instantiate object
weather_info = MeteoForecast(latitude=LOCATION[0], longitude=LOCATION[1])
````

Upon instantiation, the class also pulls the current weather data in this location, as well as its altitude. As with the MeteoHistoric class, four different metrics are recorded: 

- Temperature (```.temperature```): Air temperature in Celcius at 2 metres above the ground
- Wind Speed (```.wind```): Wind speed in km/s measured at 10 metres above ground level.
- Cloud Cover (```.cloud```): Total cloud cover as a percentage of the sky.
- Precipitation (```.precipitation```): The total precipitation (snow, rain, showers) in millimetres that has fallen in the previous hour.

You can see this from the following parameters:

````python
# Get elevation of the location, according to the OpenMeteo API
print(f"The elevation of the location is {weather_info.elevation:.1f} m")

# Get the latest wind measurement
wind_data = weather_info.wind
print(
    f"The wind was blowing at {wind_data[1][0]} km/h at {wind_data[0][0].isoformat(' ')}"
)
````

We can also obtain forecast data for the location, up to 16 days in advance. Let's try to get the forecast at the location for the upcoming week:

````python
import datetime as dt

# Get the weather the next 7 days
END = dt.datetime.now() + dt.timedelta(days=7)

# Now collect weather data using object - and defining the time
# range of interest
weather_info.get_data(end_date=END)
````

Now we have the data stored in the class, we can access it with the following properties:

- ```.temperature``` to get the temperature data
- ```.wind``` to get the wind data
- ```.cloud``` to get the cloud coverage data
- ```.precipitation``` to get the precipitation data

Each dataset is presented in a tuple, with the first element being a list of timestamps showing when each measurement was recorded, with the second element being another list - of the same size- with the numerical values of the measurement. 

````python
# Let's see how the cloud cover is expected to evolve over the next two days
cut_off = dt.datetime.now() + dt.timedelta(days=2)

# Use mask to filter data to keep only stuff less than 2 days into future
filtered_data = ((ts, data) for (ts, data) in zip(*weather_info.cloud) if ts <= cut_off)

for ts, cloud in filtered_data:
    print(f"Location is expected to be {cloud} % cloudy at {ts.isoformat(' ')}")
````
