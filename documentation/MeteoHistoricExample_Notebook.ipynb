This notebook gives some demonstration on how to use the MeteoHistorical class in order to understand how to obtain historic weather data from the Open-Meteo API. The API allows you to obtain a range of different weather variables, including temperature, wind speed, amount of cloud cover and precipitation, all of which can be accessed using this Python module. Other information - which totals an impressive 90 TB of data available from OpenMeteo - such as soil moisture, solar irradiance, dew points, etc, can be added in future if required. All of this data is available from 1st January 1940 to the current day - anywhere in the world! 

This rich source of data is perfect for training and refining your meteorological data science projects - so let's get started! The main class for accessing historical data is the ```MeteoHistorical``` class, so let's import it:

````python
from openmeteo_example import MeteoHistorical
````

Now we have imported it, let's use it! In order to create the object, we need to provide it with the latitude and longitude of a particular location. According to the [API documentation](https://open-meteo.com/en/docs/historical-weather-api#data_sources), we should expect a spatial resolution of better than 10 km, which should be more than good enough for most purposes. 
Let's choose a location in the UK, with a Latitude of 53.3557° North and 1.6303° West and pass this into the class to create the object: 

````python
# Define location
LOCATION = (53.3557, -1.6303)

# Instantiate object
weather_info = MeteoHistorical(latitude=LOCATION[0],
                               longitude=LOCATION[1])
````

Now we have create the weather_info object, we can request the following weather information for the location:

- Temperature: Air temperature in Celcius at 2 metres above the ground
- Wind Speed: Wind speed in km/s measured at 10 metres above ground level.
- Cloud Cover: Total cloud cover as a percentage of the sky.
- Precipitation: The total precipitation (snow, rain, showers) in millimetres that has fallen in the previous hour. 

These values are given on an hourly basis. We can either define two dates (as datetime objects) to designate the time range of interest. Alternatively, given these are optional values, the recent weather data is collected, with the defaults being between three weeks and two weeks for the current date of the request. Let's see this now:

````python
import datetime as dt

# Get the weather for the entirety of 2020
START = dt.datetime(2020, 1, 1)
END = dt.datetime(2020, 12, 31, 23, 59)

# Now collect weather data using object - and defining the time
# range of interest
weather_info.get_data(start_date=START,
                      end_date=END)
````

Now we have the data stored in the class, we can access it with the following properties:

- ```.temperature``` to get the temperature data
- ```.wind``` to get the wind data
- ```.cloud``` to get the cloud coverage data
- ```.precipitation``` to get the precipitation data

Each dataset is presented in a tuple, with the first element being a list of timestamps showing when each measurement was recorded, with the second element being another list - of the same size- with the numerical values of the measurement. 

Let's use this to find out what the average temperature in this location was in 2020, along with the standard deviation to get a feeling for the amount of variation:

````python
# Import statistics module from standard library - would obviously normally stick it into
# numpy but want to
import statistics

# Get data
temp_ts, temp_data = weather_info.temperature

# Calculate average value
ave_temp = statistics.mean(temp_data)
std_temp = statistics.stdev(temp_data)

print(f'The temperature at location ({weather_info.latitude:.3f}°, {weather_info.longitude:.3f}°) in {START.year} was {ave_temp:.1f} ± {std_temp:.1f} °C')
````

As stated previously, we can also get data without providing datetimes. In the case of the MeteoHistorical, it will get data from the time period between 3 weeks ago and 2 weeks ago, from the current time it is run. You can obtain this by using the ```get_data()``` method with no arguments:

````python
# Get recent data
weather_info.get_data()

# Printing out the timestamps of the last 192 measurements:
buff = weather_info.wind

print(buff[0][-192:])
````
